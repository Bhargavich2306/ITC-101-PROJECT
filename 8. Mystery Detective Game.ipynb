```python
print("🕵️ THE MISSING LAPTOP")
print("A laptop has disappeared from the college lab!")
print("You are the detective.")

score = 0

print("\nYou find three clues:")
print("1. Footprints")
print("2. A key")
print("3. A notebook")

choice = input("\nWhich clue will you investigate? ")

if choice == "1":
    print("\nThe footprints lead to the library.")
    score += 10

elif choice == "2":
    print("\nThe key belongs to the computer lab.")
    score += 20

elif choice == "3":
    print("\nThe notebook contains a suspicious name.")
    score += 30

else:
    print("Invalid choice.")

print("\nYour detective score:", score)

if score >= 30:
    print("🎉 You solved an important part of the mystery!")
elif score >= 20:
    print("🔎 You're getting closer!")
else:
    print("😅 You need more clues!")
```
