```python
campus = {
    "Gate": ["Cafeteria", "Library"],
    "Cafeteria": ["Gate", "Block A"],
    "Library": ["Gate", "Block B"],
    "Block A": ["Cafeteria", "Block B"],
    "Block B": ["Library", "Block A", "Lab"]
}

def find_route(start, destination):
    queue = [[start]]
    visited = []

    while queue:
        path = queue.pop(0)
        place = path[-1]

        if place == destination:
            return path

        if place not in visited:
            visited.append(place)

            for next_place in campus.get(place, []):
                new_path = path + [next_place]
                queue.append(new_path)

    return None

start = input("Enter starting place: ")
destination = input("Enter destination: ")

route = find_route(start, destination)

if route:
    print("Route:", " → ".join(route))
else:
    print("No route found.")
```
