```python
habits = {}

n = int(input("How many habits do you track? "))

for i in range(n):
    habit = input("Habit name: ")
    completed = int(input("Days completed this week (0-7): "))
    habits[habit] = completed

print("\n🎯 Weekly Habit Report")

for habit, days in habits.items():
    percentage = (days / 7) * 100

    print(habit, ":", days, "/ 7 days",
          "(", round(percentage, 1), "% )")

    if percentage == 100:
        print("🔥 Perfect streak!")
    elif percentage >= 70:
        print("👍 Great consistency!")
    else:
        print("💪 Keep going!")
```
