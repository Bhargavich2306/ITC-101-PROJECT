```python
subjects = int(input("Enter number of subjects: "))

marks = []

for i in range(subjects):
    mark = float(input(f"Enter marks for subject {i+1}: "))
    marks.append(mark)

average = sum(marks) / subjects

if average >= 90:
    grade = "A+"
elif average >= 80:
    grade = "A"
elif average >= 70:
    grade = "B"
elif average >= 60:
    grade = "C"
elif average >= 50:
    grade = "D"
else:
    grade = "F"

print("\nAverage Marks:", round(average, 2))
print("Grade:", grade)
```
