```python
subjects = {}
n = int(input("How many subjects? "))

for i in range(n):
    subject = input("Enter subject: ")
    hours = float(input("Hours studied: "))
    subjects[subject] = hours

total = sum(subjects.values())
average = total / n

print("\n📚 Study Report")
print("Total study hours:", total)
print("Average per subject:", round(average, 2))

for subject, hours in subjects.items():
    print(subject, ":", hours, "hours")

if average >= 5:
    print("Excellent study routine! 🌟")
elif average >= 3:
    print("Good! Keep improving. 👍")
else:
    print("Try increasing your study time. 📖")
```
