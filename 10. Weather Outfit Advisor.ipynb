```python
temperature = float(input("Enter temperature in °C: "))

if temperature < 10:
    print("🧥 Wear a warm jacket.")
elif temperature < 20:
    print("🧥 A light jacket would be comfortable.")
elif temperature < 30:
    print("👕 Light and comfortable clothes are suitable.")
else:
    print("☀️ Wear light clothes and stay hydrated.")
```
