```python
def productivity_score(tasks, completed, study_hours, distractions):
    score = 0

    score += (completed / tasks) * 50
    score += min(study_hours * 8, 30)
    score += max(20 - distractions * 5, 0)

    return round(score, 2)

tasks = int(input("Total tasks today: "))
completed = int(input("Tasks completed: "))
study_hours = float(input("Study hours: "))
distractions = int(input("Number of major distractions: "))

score = productivity_score(tasks, completed, study_hours, distractions)

print("\nYour Productivity Score:", score, "/ 100")

if score >= 80:
    print("🔥 Highly productive!")
elif score >= 50:
    print("👍 Good work!")
else:
    print("💪 You can improve tomorrow!")
```
